```python
import sympy as sym
from IPython.display import display, Math
from sympy import cos, pi

inf_bounds = (-sym.oo, sym.oo)

def latex_this(expr, apply_indent=False):
    if apply_indent:
        indent = "\\hspace{3em}"
    else:
        indent = ""
    display(Math(indent + sym.latex(expr)))
```

```python
def S(tuple_1, tuple_2, tuple_3):
    t_1, f_1 = tuple_1
    t_2, f_2 = tuple_2
    t_3, f_3 = tuple_3
    row_1 = sym.cos(4*sym.pi * ( f_1 * t_2 - f_1 * t_3 - f_2 * t_1 + f_2 * t_3 - f_3 * t_1 + f_3 * t_2))
    row_2 = sym.cos(4*sym.pi * ( f_1 * t_2 - f_1 * t_3 - f_2 * t_1 + f_2 * t_3 + f_3 * t_1 - f_3 * t_2))
    row_3 = sym.cos(4*sym.pi * ( f_1 * t_2 - f_1 * t_3 + f_2 * t_1 - f_2 * t_3 - f_3 * t_1 + f_3 * t_2))
    row_4 = sym.cos(4*sym.pi * ( f_1 * t_2 - f_1 * t_3 + f_2 * t_1 - f_2 * t_3 + f_3 * t_1 - f_3 * t_2))
    return 8*sum([row_1, row_2, row_3, row_4])
```

```python
f1, f2, f3, f4 = sym.symbols('f_1 f_2 f_3 f_4')
t1, t2, t3, t4 = sym.symbols('t_1 t_2 t_3 t_4')
```

```python

# Define the 16 terms
term1 = 8*cos(pi*(-4*f1*t2 + 4*f1*t4 - 4*f2*t1 + 4*f2*t4 - 4*f4*t1 + 4*f4*t2))
term2 = 8*cos(pi*(-4*f1*t2 + 4*f1*t4 - 4*f2*t1 + 4*f2*t4 + 4*f4*t1 - 4*f4*t2))
term3 = 8*cos(pi*(-4*f1*t2 + 4*f1*t4 + 4*f2*t1 - 4*f2*t4 - 4*f4*t1 + 4*f4*t2))
term4 = 8*cos(pi*(-4*f1*t2 + 4*f1*t4 + 4*f2*t1 - 4*f2*t4 + 4*f4*t1 - 4*f4*t2))

term5 = 8*cos(pi*(4*f1*t2 - 4*f1*t3 - 4*f2*t1 + 4*f2*t3 - 4*f3*t1 + 4*f3*t2))
term6 = 8*cos(pi*(4*f1*t2 - 4*f1*t3 - 4*f2*t1 + 4*f2*t3 + 4*f3*t1 - 4*f3*t2))
term7 = 8*cos(pi*(4*f1*t2 - 4*f1*t3 + 4*f2*t1 - 4*f2*t3 - 4*f3*t1 + 4*f3*t2))
term8 = 8*cos(pi*(4*f1*t2 - 4*f1*t3 + 4*f2*t1 - 4*f2*t3 + 4*f3*t1 - 4*f3*t2))

term9 = 8*cos(pi*(-4*f1*t3 + 4*f1*t4 - 4*f3*t1 + 4*f3*t4 + 4*f4*t1 - 4*f4*t3))
term10 = 8*cos(pi*(-4*f1*t3 + 4*f1*t4 + 4*f3*t1 - 4*f3*t4 + 4*f4*t1 - 4*f4*t3))
term11 = 8*cos(pi*(4*f1*t3 - 4*f1*t4 - 4*f3*t1 + 4*f3*t4 + 4*f4*t1 - 4*f4*t3))
term12 = 8*cos(pi*(4*f1*t3 - 4*f1*t4 + 4*f3*t1 - 4*f3*t4 + 4*f4*t1 - 4*f4*t3))

term13 = 8*cos(pi*(4*f2*t3 - 4*f2*t4 - 4*f3*t2 + 4*f3*t4 - 4*f4*t2 + 4*f4*t3))
term14 = 8*cos(pi*(4*f2*t3 - 4*f2*t4 - 4*f3*t2 + 4*f3*t4 + 4*f4*t2 - 4*f4*t3))
term15 = 8*cos(pi*(4*f2*t3 - 4*f2*t4 + 4*f3*t2 - 4*f3*t4 - 4*f4*t2 + 4*f4*t3))
term16 = 8*cos(pi*(4*f2*t3 - 4*f2*t4 + 4*f3*t2 - 4*f3*t4 + 4*f4*t2 - 4*f4*t3))

# Sum all terms
correlation_4pt = term1 + term2 + term3 + term4 + term5 + term6 + term7 + term8 + term9 + term10 + term11 + term12 + term13 + term14 + term15 + term16
```

Now build

$$ \langle S_{t_1f_1}S_{t_2f_2}S_{t_4f_4} \rangle^c + \langle S_{t_1f_1}S_{t_2f_2}S_{t_3f_3} \rangle^c +\langle S_{t_1f_1}S_{t_3f_3}S_{t_4f_4} \rangle^c+\langle S_{t_2f_2}S_{t_3f_3}S_{t_4f_4} \rangle^c$$


```python
sum_of_3pts = S((t1, f1),(t2, f2),(t4, f4)) + S((t1, f1),(t2, f2),(t3, f3)) + S((t1, f1),(t3, f3),(t4, f4)) + S((t2, f2),(t3, f3),(t4, f4))
```

```python
latex_this((correlation_4pt - sum_of_3pts).simplify())
```
